```python
import re

def extract_text_between_brackets(input_string):
    pattern = r'\[(.*?)\]'
    matches = re.findall(pattern, input_string)
    return matches

# Example usage:
input_string = "This is [some text] between square brackets]."
result = extract_text_between_brackets(input_string)
print(result)
```
